```python
# Sales Trend Over Time Monthly & Yearly
monthly_sales=data.groupby(['order date'],as_index=False).sum()
monthly_sales
```

```python
monthly_sales = monthly_sales.set_index('order date')
monthly_sales.index = pd.to_datetime(monthly_sales.index)
monthly_sales = monthly_sales.resample('ME').sum()
monthly_sales
```

```python
plt.figure(figsize=(25,10))
plt.plot(monthly_sales['sales'])
plt.xlabel("order date")
plt.ylabel("sales")
plt.title("Monthly Sales Trend")
plt.show()
```

```python
year_sales = monthly_sales.resample('YE').sum()
plt.figure(figsize=(25,10))
plt.plot(year_sales['sales'])
plt.xlabel(['order date'])
plt.xlabel(['sales'])
plt.title("yearly Sales Trend")
```

```python
monthly_sales = data.groupby(['order date'], as_index=False).sum()

# Set the Order Date column as the index of the dataframe
monthly_sales = monthly_sales.set_index('order date')
monthly_sales.index = pd.to_datetime(monthly_sales.index)

monthly_sales = monthly_sales.resample('ME').sum() # M for month

# Plot
plt.figure(figsize=(25,8))
plt.plot(monthly_sales['profit'])
plt.xlabel("order date")
plt.ylabel("profit")
plt.title("Monthly Profit Trend")
plt.show()
```

```python
year_sales = monthly_sales.resample('YE').sum()
plt.figure(figsize=(25,10))
plt.plot(year_sales['profit'])
plt.xlabel(['order date'])
plt.xlabel(['profit'])
plt.title("yearly Profit Trend")
```

```python
# Which Place Generates The HIghest Sales & Profit
places = data[['country','city','state','region']]
places
```

```python
for place in places.columns:
    print(place,':',places[place].nunique())
```

```python
places = data[['country','city','state','region','sales','profit']]
```

```python
# Country
places_group= places.groupby(['country']).sum(['sales','profit'])
places_group
```

```python
# city
city_group= places.groupby(['city'],as_index=False).sum(['sales','profit'])
city_group
city_group.sort_values(by='profit', ascending=False, inplace=True)

top_city=pd.DataFrame(city_group[:5])
top_city
```

```python
plt.figure(figsize=(15,10))
plt.bar(top_city['city'], top_city['profit'], align='center')
plt.xlabel("City")
plt.ylabel("Profit")
plt.title("Top 5 Cities by Profit")
plt.xticks(rotation=90)

plt.show()
```

```python
# Region
region_group= places.groupby(['region'],as_index=False).sum(['sales','profit'])
region_group

region_group.sort_values(by='sales', ascending=False, inplace=True)
top_region=pd.DataFrame(region_group[:5])
top_region

plt.figure(figsize=(15,10))
plt.bar(top_region['region'], top_region['profit'], align='center')
plt.xlabel("region")
plt.ylabel("profit")
plt.title("Best Region by Profit")
plt.xticks(rotation=90)

plt.show()
```

```python
# States
grouped_data = places.groupby(['state'], as_index=False).sum()
grouped_data.sort_values(by='sales', ascending=False, inplace=True)
plt.figure(figsize=(25,15))
plt.bar(grouped_data['state'], grouped_data['sales'], align='center',)
plt.xlabel("State")
plt.ylabel("Sales")
plt.title("Sales Generated by State")
plt.xticks(rotation=90)
plt.show()
```

```python
data['ROS']=data['profit'] / data['sales']
data
```

```python
data['ROS%']=round(data.ROS*100,2)
```

```python
ros_group= data.groupby('category')['ROS'].mean()
ros_group.plot(kind='bar')

plt.title("Average Return of sales by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Average Profit Margin")

plt.show()
```
